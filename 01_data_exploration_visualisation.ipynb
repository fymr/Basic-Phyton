```python
activity_dict = dict(zip(activity_labels[0],activity_labels[1]))
y_train["activity"] = y_train[0].map(activity_dict)
y_train.head()
```

```python
df = pd.DataFrame(
    {
        "feature": x_train["tBodyAcc-mean()-X"],
        "activity": y_train["activity"]

    }
)
df
```

```python
plt.boxplot(df[df["activity"]== "WALKING"]["feature"])
plt.title("walking")
plt.show()
```

```python

walking = df[df["activity"] == "WALKING"]["feature"]
walking_upstairs = df[df["activity"] == "WALKING_UPSTAIRS"]["feature"]
walking_downstairs = df[df["activity"] == "WALKING_DOWNSTAIRS"]["feature"]
sitting = df[df["activity"] == "SITTING"]["feature"]
standing = df[df["activity"] == "STANDING"]["feature"]
laying = df[df["activity"] == "LAYING"]["feature"]

plt.boxplot([
    walking,
    walking_upstairs,
    walking_downstairs,
    sitting,
    standing,
    laying
])

plt.xlabel("Activity")
plt.ylabel("Feature : tBodyAcc-mean()-X")
plt.title("Feature Distribution by Activity")

plt.show()
```

```python
df= pd.DataFrame(
    {
        "feature" :x_train["tBodyAcc-std()-X"],
        "activity":y_train["activity"]
    }
)

walking = df[df["activity"] == "WALKING"]["feature"]
walking_upstairs = df[df["activity"] == "WALKING_UPSTAIRS"]["feature"]
walking_downstairs = df[df["activity"] == "WALKING_DOWNSTAIRS"]["feature"]
sitting = df[df["activity"] == "SITTING"]["feature"]
standing = df[df["activity"] == "STANDING"]["feature"]
laying = df[df["activity"] == "LAYING"]["feature"]

plt.boxplot([
    walking,
    walking_upstairs,
    walking_downstairs,
    sitting,
    standing,
    laying
])

plt.xlabel("Activity")
plt.ylabel("Feature : tBodyAcc-std()-X")
plt.title("Feature Distribution by Activity")

plt.show()
```

```python
plt.hist(x_train["tBodyAcc-std()-X"], bins=30, color="blue")
plt.xlabel("tBodyAcc-std()-X")
plt.ylabel("frequency")
plt.title("Histogram for tBodyAcc-std()-X")
plt.show()
```

```python
plt.scatter(
    x_train["tBodyAcc-std()-X"],
    x_train["tBodyAcc-mean()-X"],
    alpha=0.5
)
plt.xlabel("tBodyAcc-std()-X")
plt.ylabel("tBodyAcc-mean()-X")
plt.title("Scatter Plot of Standart deviation and mean")
plt.show()
```

```python

df = pd.DataFrame(
    {
        "mean_x" : x_train["tBodyAcc-mean()-X"],
        "std_x" : x_train["tBodyAcc-std()-X"],
        "activity" : y_train["activity"]
    }
)

walking = df[df["activity"] == "WALKING"]
plt.scatter(
    walking["mean_x"],
    walking["std_x"]

)
plt.xlabel("mean_x")
plt.ylabel("std_x")
plt.title("walking")
plt.show()
```

```python

df = pd.DataFrame(
    {
        "mean_x" : x_train["tBodyAcc-mean()-X"],
        "std_x" : x_train["tBodyAcc-std()-X"],
        "activity" : y_train["activity"]
    }
)

walking = df[df["activity"] == "WALKING"]
sitting =df[df["activity"]=="SITTING"]
plt.scatter(
    walking["mean_x"],
    walking["std_x"],
    color="blue"
)
plt.scatter(
    sitting["mean_x"],
    sitting["std_x"],
    color = "red"
)
plt.xlabel("mean_x")
plt.ylabel("std_x")
plt.title("walking vs sitting")
plt.show()
```
