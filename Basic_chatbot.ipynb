```python
bot_name: str = 'Bob'
print(f'Hello! I\'m {bot_name}, how can I help you?')

while True:
  user_input: str= input('You: ').lower()

  if user_input in['hi','hello']:
    print(f'{bot_name}: Hi! How can I help you?')

  elif user_input in['bye', 'goodbye']:
    print(f' {bot_name}: Goodbye!')

  elif user_input in['+','add'] :
    print(f'{bot_name}: Sure ! Please give me two numbers.')
    try:
      num1 : float = float(input('First number: '))
      num2 : float = float(input('Second number: '))
      print(f'{bot_name}: {num1} + {num2} = {num1 + num2}')

    except ValueError:
      print(f'{bot_name}: It is not a valid number. Try again')

  else:
   print(f'{bot_name}: I\'m sorry , I don\'t understand that')

```
